```python
# Create a list named name_list with five string elements.  
name_list=['Kevin','Anna','Nate','Rowan','Tom']
name_list
```

```python
# Append the string "Yazhini" to name_list. 
print(name_list)
name_list.append('Yazhini')
print(name_list)
```

```python
#  Insert the element 30 at index 2 in age_list. 
age_list.insert(2,30)
print(age_list)
```

```python
# Remove the string "Yazhini" from name_list.
name_list.remove('Yazhini')
name_list
```

```python
# Pop the last element from age_list
age_list.pop()
age_list
```

```python
# Extend the age_list with additional ages [29, 30, 26].

age_list.extend([29,30,26])
print(age_list)
```

```python
# Sort age_list in descending order.
age_list.sort(reverse=True)
age_list
```

```python
# Find Max age, Min age and sum of all ages from age_list.
max_age=max(age_list)
min_age=min(age_list)
sum_age=sum(age_list)
print(max_age)
print(min_age)
print(sum_age)
```

```python
''' Create two sets:  
set1 with values: {1, 3, 5, 7, 9}  
set2 with values: {2, 3, 5, 8, 10} '''
set1={1,3,5,7,9}
set2={2,3,5,8,10}
print(set1)
print(set2)
```

```python
# Compute and print the union and intersection of set1 and set2.
set1={1,3,5,7,9}
set2={2,3,5,8,10}
new=set1.union(set2)
new
```

```python
# intersection
set1={1,3,5,7,9}
set2={2,3,5,8,10}
c=set1.intersection(set2)
c
```

```python
# Operators & Conditional Statements
score=int(input("Enter your score"))
if 0<=score>=10:
    print("Invalid input. Please enter a score between 0 and 10.")
else:
    if score>7:
        print("Above Average:Keep up the good work.")
    elif 4<=score<=7:
        print("Average: Good effort! Keep practicing, there's room for improvement.")
    else:
        print("Below Average: Need to improve your performance")
```
